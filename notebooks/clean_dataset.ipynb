## Emma

- media_type: filled the 80 missing values with UNKNOWN.

## Timothy

- post_avg_NRx: capped 19 high values at Q3 + 1.5 × IQR of the raw data.
- pre_NRx_avg: capped 14 high values at Q3 + 1.5 × IQR of the raw data.
- npi_specialty: renamed 3320 - Anesthesiologists to Anesthesiology.
- region: renamed West South Central to Southwest.
- NRx_lift: recalculated 105 values as post_NRx minus pre_NRx, rounded to 2 decimals.

```python
for col in ["post_avg_NRx", "pre_NRx_avg"]:
    q1 = raw[col].quantile(0.25)
    q3 = raw[col].quantile(0.75)
    upper = q3 + 1.5 * (q3 - q1)
    df[col] = df[col].clip(upper=upper)

df["npi_specialty"] = df["npi_specialty"].replace("3320 - Anesthesiologists", "Anesthesiology")
df["region"] = df["region"].replace("West South Central", "Southwest")
df["NRx_lift"] = (df["post_NRx"] - df["pre_NRx"]).round(2)
```

## Nida

- total_spend: no change, its 3 missing values were in the rows Emma dropped.
- actual_spend_per_engagement: no change, no missing values.
- spend_per_NRx_lift: no change, no missing values, even though NRx_lift was recalculated.

```python
df["cpe"] = df["cpe"].fillna((df["total_budget_spent"] / df["total_engagements"]).round(2))
df.isna().sum().sum()
```
