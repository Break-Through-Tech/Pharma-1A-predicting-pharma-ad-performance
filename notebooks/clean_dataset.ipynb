## Suggested additions (Rhea)

These steps build on the team's agreed cleaning above. They do not undo any team decision. They fill the gaps that came up in review so the file is ready for modeling:

1. **Profession labels:** one consistent format.
2. **Start date:** converted to a real date, with time features added.
3. **NRx columns:** `NRx_lift_pct` and `spend_per_NRx_lift` recalculated so they match the recalculated `NRx_lift`.
4. **Flags:** mark rows that would distort scoring or were filled in, instead of removing them.
5. **Encoding and feature list:** a model-ready file plus the exact columns each target may use.

### 1. Profession labels

After the advisor's code list is applied, the column mixes title case names, lowercase text and 16 bare numbers. Text labels are put in title case, and codes that are not on the list become `Unmapped code N` so they are clearly labels, not numbers.

**To confirm with the advisor:** the code list includes categories such as "Aboriginal and Torres Strait Islander Health/Mental Health Worker", which look like a mental health workforce list. It is worth checking that this is the right list for WebMD's HCP data.

```python
def tidy_profession(v):
    if isinstance(v, (int, float)) or str(v).isdigit():
        return f"Unmapped code {int(v)}"
    if v in ("UNKNOWN", "Not stated"):
        return v
    return str(v).title() if str(v).islower() else v

before = df["profession"].copy()
df["profession"] = df["profession"].map(tidy_profession)
print("Rows relabeled:", (before.astype(str) != df["profession"]).sum())
df["profession"].value_counts().to_frame("rows")
```

### 2. Start date and time features

`campaign_start_date` was still text with a time and " UTC" on the end. It becomes a plain date, and four time features are added. `campaign_age_days` counts back from the latest start date in the data, so the value is the same whenever the notebook is run.

```python
start = pd.to_datetime(df["campaign_start_date"].str.replace(" UTC", "", regex=False))
df["campaign_start_date"] = start.dt.strftime("%Y-%m-%d")
df["start_year"] = start.dt.year
df["start_quarter"] = start.dt.quarter
df["start_month"] = start.dt.month
REFERENCE_DATE = start.max().normalize()
df["campaign_age_days"] = (REFERENCE_DATE - start.dt.normalize()).dt.days
print("Reference date for campaign_age_days:", REFERENCE_DATE.date())
df[["campaign_start_date", "start_year", "start_quarter", "start_month", "campaign_age_days"]].head()
```

### 3. Keep the NRx columns consistent

Timothy recalculated `NRx_lift` as `post_NRx - pre_NRx`. Two columns are built from it:

- `NRx_lift_pct = NRx_lift / pre_NRx x 100` (4 decimals)
- `spend_per_NRx_lift = total_spend / NRx_lift` (2 decimals)

Only the rows that no longer match (beyond normal rounding) are recalculated. Every other value is left exactly as it was in the raw file.

```python
new_pct = (df["NRx_lift"] / df["pre_NRx"] * 100).round(4)
new_spl = (df["total_spend"] / df["NRx_lift"]).round(2)
fix_pct = (df["NRx_lift_pct"] - new_pct).abs() > 0.01
fix_spl = (df["spend_per_NRx_lift"] - new_spl).abs() > 0.001 * new_spl.abs()
df.loc[fix_pct, "NRx_lift_pct"] = new_pct[fix_pct]
df.loc[fix_spl, "spend_per_NRx_lift"] = new_spl[fix_spl]
print("NRx_lift_pct rows recalculated:", fix_pct.sum())
print("spend_per_NRx_lift rows recalculated:", fix_spl.sum())
df.loc[fix_pct | fix_spl, ["pre_NRx", "post_NRx", "NRx_lift", "NRx_lift_pct", "total_spend", "spend_per_NRx_lift"]]
```

### 4. Flags

Rows are marked rather than removed, so every campaign stays in the file.

| Flag | Meaning | Why it matters |
|---|---|---|
| `cpe_imputed` | cpe was filled from budget / engagements | Shows which cpe values were calculated |
| `spend_near_zero` | total_spend under $1,000 | MAPE breaks at $0 and explodes near it, so these rows are left out of MAPE |
| `budget_anomaly` | Budget is $0, does not match cpe x engagements, or gives a cpe under $1 | Leave out of cost-efficiency analysis |
| `targeting_info_missing` | Country, pillar, media type and profession were all missing in the raw data | The missing block is a pattern the model can learn from |

```python
src = raw.loc[df.index]
df["cpe_imputed"] = src["cpe"].isna().astype(int)
df["spend_near_zero"] = (df["total_spend"] < 1000).astype(int)
mismatch = (df["cpe"] * df["total_engagements"] - df["total_budget_spent"]).abs() > 1
df["budget_anomaly"] = ((df["total_budget_spent"] == 0) | mismatch | (df["cpe"] < 1)).astype(int)
df["targeting_info_missing"] = src[["country", "pillar", "media_type", "profession"]].isna().all(axis=1).astype(int)

flags = ["cpe_imputed", "spend_near_zero", "budget_anomaly", "targeting_info_missing"]
df[flags].sum().to_frame("rows flagged")
```

### 5. Encoded file and feature list

**Encoded file (`data/cleaned_df_encoded.csv`)**

- Log versions of the heavily skewed columns are added, including `log_total_spend`, the version of the target we model.
- Every category column is one-hot encoded. None of them has a natural order, so turning them into numbers would invent a ranking.
- Categories with fewer than 5 rows are grouped into `OTHER`, so we do not create columns that are almost always zero.
- `product` becomes a single `has_named_product` flag, because most products appear only once or twice.
- Raw text columns (`brand`, `country_name`, `product`, `campaign_start_date`) are left out.
- Scaling is left to the modeling pipeline so it is fit on training data only.

**Feature list (`reports/feature_sets.json`)**

Columns that are calculated from the target, or only known after a campaign ends, are never features. For the CPE target, budget is also left out, because budget / engagements = cpe.

```python
import json
import numpy as np

enc = df.copy()
for c in ["unique_reach", "unique_engaged_hcps", "total_engagements", "total_budget_spent", "total_spend"]:
    enc[f"log_{c}"] = np.log1p(enc[c])
enc["has_named_product"] = (enc["product"] != "unspecified").astype(int)

ONEHOT_COLS = ["company", "channel_tag", "sub_product_name", "country", "pillar", "media_type",
               "profession", "npi_specialty", "region"]
for c in ONEHOT_COLS:
    counts = enc[c].value_counts()
    enc[c] = enc[c].where(~enc[c].isin(counts[counts < 5].index), "OTHER")

enc = enc.drop(columns=["brand", "country_name", "product", "campaign_start_date"])
enc = pd.get_dummies(enc, columns=ONEHOT_COLS, prefix=ONEHOT_COLS, dtype=int)
enc.columns = enc.columns.str.replace(r"[^0-9a-zA-Z_]+", "_", regex=True).str.strip("_")
assert enc.columns.is_unique

NEVER_FEATURES = {"total_spend", "log_total_spend", "actual_spend_per_engagement", "spend_per_NRx_lift",
                  "post_NRx", "post_avg_NRx", "NRx_lift", "NRx_lift_pct", "spend_near_zero"}
CPE_EXTRA_EXCLUDE = {"cpe", "total_budget_spent", "log_total_budget_spent", "cpe_imputed", "budget_anomaly"}
spend_features = [c for c in enc.columns if c not in NEVER_FEATURES]
cpe_features = [c for c in spend_features if c not in CPE_EXTRA_EXCLUDE]

feature_sets = {
    "notes": "Column names refer to data/cleaned_df_encoded.csv. Generated by notebooks/clean_dataset.ipynb.",
    "total_spend": {
        "role": "primary target",
        "target_column": "log_total_spend",
        "report_on_scale": "total_spend (convert predictions back with np.expm1)",
        "metrics": ["MAPE (goal under 30%)", "R2 (goal above 0.55)"],
        "train_rows": "all rows",
        "mape_rows": "spend_near_zero == 0",
        "features": spend_features,
    },
    "cpe": {
        "role": "secondary target (confirm meta_mapped_cpe with Challenge Advisor)",
        "target_column": "cpe",
        "train_rows": "all rows",
        "features": cpe_features,
    },
    "never_features": sorted(NEVER_FEATURES),
}
print(f"Encoded file: {enc.shape[0]} rows x {enc.shape[1]} columns")
print(f"total_spend features: {len(spend_features)}  |  cpe features: {len(cpe_features)}")
```

### Validation checks

```python
checks = {
    "414 rows in both files": len(df) == len(enc) == 414,
    "no blanks in cleaned_df": df.isna().sum().sum() == 0,
    "no blanks in encoded file": enc.isna().sum().sum() == 0,
    "encoded file is all numbers": all(pd.api.types.is_numeric_dtype(t) for t in enc.dtypes),
    "profession has no bare numbers": not df["profession"].astype(str).str.fullmatch(r"\d+").any(),
    "start dates are real dates": pd.to_datetime(df["campaign_start_date"], errors="coerce").notna().all(),
    "NRx_lift = post_NRx - pre_NRx": ((df["post_NRx"] - df["pre_NRx"]).round(2) - df["NRx_lift"]).abs().max() < 0.005,
    "NRx_lift_pct matches NRx_lift": ((df["NRx_lift"] / df["pre_NRx"] * 100) - df["NRx_lift_pct"]).abs().max() <= 0.01,
    "spend_per_NRx_lift matches NRx_lift": ((df["total_spend"] / df["NRx_lift"]).round(2) - df["spend_per_NRx_lift"]).abs().le(
        0.001 * df["spend_per_NRx_lift"].abs()).all(),
    "cpe x engagements = budget (non-anomaly rows)": (
        (df["cpe"] * df["total_engagements"] - df["total_budget_spent"]).abs()[df["budget_anomaly"] == 0] <= 1).all(),
    "no leaking column in spend features": not (set(spend_features) & NEVER_FEATURES),
    "no budget or leaking column in cpe features": not (set(cpe_features) & (NEVER_FEATURES | CPE_EXTRA_EXCLUDE)),
}
assert all(checks.values()), "A validation check failed"
pd.Series(checks, name="passes").to_frame()
```
